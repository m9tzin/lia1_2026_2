# Detector de Objetos com PyTorch + YOLO11 + ONNX

## Objetivo

Utilizar um modelo YOLO11n baseado em PyTorch, pré-treinado no COCO, realizar detecção de objetos com **bounding boxes**, apresentar as **80 classes em português** e exportar o detector para **ONNX**.

Fluxo:

**Imagem → YOLO11/PyTorch → Detecção → Classe em português + Confiança + Bounding Box → ONNX**

> Observação: traduzir os nomes das classes não altera o conhecimento aprendido pelo modelo. O detector continua usando os IDs das 80 classes COCO; neste notebook associamos esses IDs aos respectivos nomes em português.

```python
import torch
import ultralytics
import onnx
import onnxruntime as ort

print("PyTorch:", torch.__version__)
print("Ultralytics:", ultralytics.__version__)
print("CUDA disponível:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

## 2. As 80 classes COCO em português

A ordem deve permanecer exatamente igual à ordem original do COCO, pois o índice é a classe utilizada pelo modelo.

```python
# Classes COCO em português
classes_pt = """
pessoa,bicicleta,carro,motocicleta,avião,ônibus,trem,caminhão,barco,semáforo,
hidrante,placa de pare,parquímetro,banco,pássaro,gato,cachorro,cavalo,ovelha,vaca,
elefante,urso,zebra,girafa,mochila,guarda-chuva,bolsa,gravata,mala,frisbee,
esquis,snowboard,bola esportiva,pipa,taco de beisebol,luva de beisebol,skate,
prancha de surfe,raquete de tênis,garrafa,taça,xícara,garfo,faca,colher,tigela,
banana,maçã,sanduíche,laranja,brócolis,cenoura,cachorro-quente,pizza,rosquinha,
bolo,cadeira,sofá,planta em vaso,cama,mesa de jantar,vaso sanitário,televisão,
notebook,mouse,controle remoto,teclado,celular,micro-ondas,forno,torradeira,pia,
geladeira,livro,relógio,vaso,tesoura,urso de pelúcia,secador de cabelo,escova de dentes
"""
COCO_PT = dict(enumerate(
    [c.strip() for c in classes_pt.replace("\\n", "").split(",")]
))
print(f"{len(COCO_PT)} classes carregadas.")
```

## 3. Carregar o YOLO11n

O `yolo11n.pt` já é pré-treinado no conjunto COCO com as 80 classes.

Por isso, **não é necessário retreinar todo o COCO apenas para traduzir os nomes das classes**.

O modelo é implementado sobre PyTorch e será baixado automaticamente na primeira execução.

```python
from ultralytics import YOLO

modelo = YOLO("yolo11n.pt")

print("Modelo carregado.")
print("Número de classes:", len(modelo.names))
```

## 4. Realizar uma inferência

Usaremos uma imagem pública de exemplo para verificar o funcionamento do detector.

A confiança mínima inicial será de **25%**.

```python
imagem = "https://ultralytics.com/images/bus.jpg"

resultados = modelo.predict(
    source=imagem,
    conf=0.25,
    save=False
)

resultado = resultados[0]
resultado.show()
```

## 5. Mostrar classes, confianças e bounding boxes em português

O modelo retorna o ID numérico da classe. Usamos esse ID para consultar `COCO_PT`.

```python
r = resultados[0]

if len(r.boxes) == 0:
    print("Nenhum objeto detectado.")
else:
    for i, box in enumerate(r.boxes):
        classe_id = int(box.cls[0])
        confianca = float(box.conf[0])
        x1, y1, x2, y2 = box.xyxy[0].tolist()

        print(f"Objeto {i + 1}")
        print("Classe:", COCO_PT[classe_id])
        print(f"Confiança: {confianca:.2%}")
        print(
            f"Bounding box: "
            f"[{x1:.1f}, {y1:.1f}, {x2:.1f}, {y2:.1f}]"
        )
        print("-" * 45)
```

## 6. Gerar uma imagem com bounding boxes e rótulos em português

Esta etapa desenha as caixas diretamente sobre a imagem usando OpenCV.

```python
import cv2
import matplotlib.pyplot as plt
from ultralytics.utils.plotting import colors

imagem_original = r.orig_img.copy()

for box in r.boxes:
    classe_id = int(box.cls[0])
    confianca = float(box.conf[0])
    x1, y1, x2, y2 = map(int, box.xyxy[0].tolist())

    cor = colors(classe_id, bgr=True)
    rotulo = f"{COCO_PT[classe_id]} {confianca:.0%}"

    cv2.rectangle(imagem_original, (x1, y1), (x2, y2), cor, 2)

    (largura, altura), _ = cv2.getTextSize(
        rotulo, cv2.FONT_HERSHEY_SIMPLEX, 0.6, 2
    )

    y_texto = max(y1, altura + 10)

    cv2.rectangle(
        imagem_original,
        (x1, y_texto - altura - 10),
        (x1 + largura + 6, y_texto),
        cor,
        -1
    )

    cv2.putText(
        imagem_original,
        rotulo,
        (x1 + 3, y_texto - 5),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (255, 255, 255),
        2
    )

plt.figure(figsize=(12, 8))
plt.imshow(cv2.cvtColor(imagem_original, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("Detecção de Objetos — COCO80 em Português")
plt.show()
```

## 7. Exportar o detector para ONNX

Agora exportaremos o YOLO11n para ONNX.

O arquivo gerado poderá ser utilizado em outras aplicações, inclusive em aplicações web com **ONNX Runtime Web**.

```python
caminho_onnx = modelo.export(
    format="onnx",
    imgsz=640,
    simplify=True,
    opset=17
)

print("ONNX gerado em:", caminho_onnx)
```

## 8. Validar o arquivo ONNX

Esta etapa verifica se a estrutura do arquivo ONNX é válida e mostra suas entradas e saídas.

```python
modelo_onnx = onnx.load(caminho_onnx)
onnx.checker.check_model(modelo_onnx)

print("✓ Estrutura ONNX válida")

sessao = ort.InferenceSession(
    caminho_onnx,
    providers=["CPUExecutionProvider"]
)

print("\nENTRADAS")
for entrada in sessao.get_inputs():
    print("Nome:", entrada.name)
    print("Shape:", entrada.shape)
    print("Tipo:", entrada.type)

print("\nSAÍDAS")
for saida in sessao.get_outputs():
    print("Nome:", saida.name)
    print("Shape:", saida.shape)
    print("Tipo:", saida.type)
```

## 9. Salvar a lista de classes em português

Como um arquivo ONNX não deve ser tratado como garantia de transporte dos nomes das classes, também geraremos um arquivo `classes_coco80_pt.txt`.

Ele poderá ser utilizado pela aplicação de inferência para mapear:

**ID da classe → nome em português**

```python
arquivo_classes = "classes_coco80_pt.txt"

with open(arquivo_classes, "w", encoding="utf-8") as f:
    for i in range(80):
        f.write(COCO_PT[i] + "\n")

print("Arquivo criado:", arquivo_classes)
```
